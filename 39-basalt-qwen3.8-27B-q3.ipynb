# Running `basalt` as a server with a quantized model and multi-token-prediction

Now, we will try yet another software [`basalt`](https://github.com/jesdga95/basalt).
`basalt` is optimized for *RTX 5090* and makes optimal use of the Blackwell architecture. 

We will use the OpenAI client:

````python
from openai import OpenAI
````

An API key can be added, but we don't need it here

````python
openai_api_key = "EMPTY"
openai_api_base = "http://localhost:9090/v1"
client = OpenAI(
    api_key=openai_api_key,
    base_url=openai_api_base,
)
````

Start `basalt` with the command `bin/basalt serve --config qwen-abliterated.json --port 9090 --host 0.0.0.0 --max-context 131072 --vision dynamic`. `qwen-abliterated.json` should contain the following data:
```json
{
  "engine": {
    "expert-cache": "auto",
    "prefill": "auto",
    "resident-experts": true
  },
  "models": [
    {
      "gpu": "0",
      "kv": "int8",
      "model": "models/Qwen3.8-FN-GSQ-RCO-abliterated-IQ3_S-Q8PLE.basalt"
    }
  ],
  "server": {
    "port": 9090
  },
  "version": 2
}
```

You have to download the model separately from https://huggingface.co/Shockem/Qwen3.8-Flash-Next-GSQ-RCO-abliterated-IQ3_S-Q8PLE-Basalt

````python
model = "Qwen3.8-Flash-Next GSQ-RCO abliterated IQ3_S"
````

````python
%%time
completion = client.chat.completions.create(
    model=model, 
    messages=[{ "role": "user",
                "content": "Erkläre den Heise Verlag!" } ]
)
````

Note the speed of the generation (in the terminal window). This is also very
fast, even if it cannot always reach the speed of `exl3`, especially with many
parallel requests.

````python
from IPython.display import display, Markdown
display(Markdown(completion.choices[0].message.content))
````

The `responses` API is also support by `ninfer`. Let's generate `python` code, which supposedly can also
be created by the draft model:

````python
%%time
response = client.responses.create(model=model, 
                                   input="Write a quicksort in Python")
````

````python
display(Markdown(response.output_text))
````
